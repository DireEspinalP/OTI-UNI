# Estructuras de Datos

Una **estructura de datos** es un modo de organizar la información en la memoria de la
computadora, implementado en un lenguaje de programación específico.

| Lenguaje | Estructuras de datos principales |
|---|---|
| **C** | Arrays, `struct`, `union`, listas enlazadas, pilas, colas, árboles, tablas hash |
| **C++** | Contenedores secuenciales (`array`, `vector`, `deque`, `list`, `forward_list`), asociativos (`set`, `map`, ...), adaptadores |
| **Julia** | Arreglos (vectores, matrices, dispersos), diccionarios, conjuntos, tuplas, structs, named tuples |
| **Python** | Listas, tuplas, diccionarios, sets, módulo `collections` (`deque`, `namedtuple`, `Counter`, `defaultdict`) |

En este notebook trabajamos con las estructuras de datos nativas de **Python**
(`list`, `tuple`), además de dos módulos muy usados: `datetime` y `numpy`.


## 1. El tipo de dato `list`

### 1.1 Definición
Una lista es una secuencia **ordenada** y **mutable** que puede mezclar cualquier tipo de dato.

```python
# Una lista puede contener elementos de distintos tipos, incluida otra lista
lista1 = []
lista2 = [12, 12.3, "Trece", 12 + 6j, True, lista1]

print(type(lista1), type(lista2))
```

### 1.2 Acceso a los elementos: indexación y slicing

- De **izquierda a derecha**: índices `0, 1, 2, 3, ...`
- De **derecha a izquierda**: índices `-1, -2, -3, ...`
- La notación `slice` (`inicio:fin:paso`) permite extraer sub-listas.

```python
segunda_componente = lista2[1]          # segundo elemento
antepenultima = lista2[-3]              # antepenúltimo elemento
primeras_tres = lista2[:3]              # tres primeros elementos
ultimos_dos = lista2[-2:]               # últimos dos elementos

print("Segunda componente :", segunda_componente)
print("Antepenúltima      :", antepenultima)
print("Primeras tres       :", primeras_tres)
print("Últimos dos          :", ultimos_dos)
```

### 1.3 Concatenación (`+`) y repetición (`*`)

```python
lista3 = lista2 + lista2[-3:]   # concatenación
lista4 = 2026 * lista3          # repetición
lista5 = 19 * lista3 + lista4 * 82

for l in (lista1, lista2, lista3, lista4, lista5):
    print(type(l), "- longitud:", len(l))
```

### 1.4 Funciones básicas: `len`, `min`, `max`, `sum`

Generamos una lista de números pseudoaleatorios para probarlas.

```python
import random as rnd

datos = [rnd.randint(0, 20) for _ in range(15)]   # list comprehension
print("Datos       :", datos)
print("Longitud    :", len(datos))
print("Mínimo      :", min(datos))
print("Máximo      :", max(datos))
print("Suma        :", sum(datos))
```

### 1.5 Métodos de una lista

Los métodos de interés de un objeto `list` son:

`append`, `clear`, `copy`, `count`, `extend`, `index`, `insert`, `pop`, `remove`, `reverse`, `sort`

(se pueden consultar todos con `dir(lista)` o la ayuda de cada método con `help(lista.metodo)`).

#### `append` — agrega un elemento al final

```python
lista1 = []
lista1.append(19)
lista1.append("PIT705 : ejemplo de append")
print(lista1)
```

#### `clear` — vacía la lista

```python
lista1.clear()
print("Después de clear():", lista1)

del lista1   # elimina por completo la variable de memoria
```

#### `copy` — copia independiente vs. asignación simple

```python
# Asignación simple: ambas variables apuntan a la MISMA lista en memoria
datos1 = datos
datos1[-2] = 0
print("datos1 (modificada):", datos1)
print("datos  (también cambió):", datos)
```

```python
# Con copy() se crea una lista independiente
datos = [rnd.randint(0, 20) for _ in range(15)]   # reiniciamos datos
datos_copy = datos.copy()
datos_copy[-1] = 666
datos_copy[3] = 23

print("Copia      :", datos_copy)
print("Originales :", datos)
```

#### `count` — número de repeticiones de un valor

```python
lista7 = [rnd.randint(16, 20) for _ in range(200)]
print("Longitud       :", len(lista7))
print("Repeticiones 19:", lista7.count(19))
print("Repeticiones 0 :", lista7.count(0))
```

#### `extend` — agrega los elementos de un iterable (str, range, list, ...)

```python
lista8 = []
lista8.extend("ARRIBA PERU")     # un str es iterable: agrega letra por letra
print(lista8)

lista9 = list()                   # otra forma de crear una lista vacía
lista9.extend(range(12, 66, 17))
print(lista9)
```

#### `index` — posición de la primera ocurrencia de un valor

```python
print(lista8.index("R"))

try:
    lista8.index(666)          # error si el valor no existe
except ValueError as e:
    print("Error esperado:", e)
```

#### `insert` — inserta un elemento en una posición dada

```python
lista8.insert(0, "!!!")   # inserta antes del primer elemento
print(lista8)
```

#### `pop` — remueve y devuelve un elemento (por defecto el último)

```python
print("Antes:", datos)
ultimo = datos.pop()
print("pop() devolvió:", ultimo)
print("Después:", datos)
```

#### `remove` — elimina la primera ocurrencia de un valor

```python
for _ in range(lista7.count(19)):
    lista7.remove(19)

print("Repeticiones de 19 tras remove():", lista7.count(19))
```

#### `reverse` — invierte el orden de la lista in-place

```python
print("Antes  :", lista2)
lista2.reverse()
print("Después:", lista2)
```

#### `sort` — ordena la lista in-place

```python
lista10 = [rnd.randint(0, 20) for _ in range(10)]
print("Antes  :", lista10)
lista10.sort()
print("Después:", lista10)
```

## 2. El tipo de dato `tuple`

### 2.1 Definición e inmutabilidad

Una tupla es como una lista, pero **inmutable**: una vez creada no se pueden
modificar sus elementos.

```python
t1 = ()
t2 = (12, 34, 56, 78, "AAA", "19")
t3 = ([], [12, 45], True, list(), [1 for _ in range(5)])

for t in (t1, t2, t3):
    print(type(t), t)
```

```python
# Los datos de tipo tupla son inmutables -> esto genera un error
try:
    t2[2] = "666"
except TypeError as e:
    print("Error esperado:", e)
```

### 2.2 Concatenación y repetición
Al igual que con `str` y `list`, las tuplas admiten `+` y `*`.

```python
t4 = t2 + t3       # concatenación
t5 = 15 * t3        # repetición

print(type(t4), t4)
print(type(t5), len(t5))

# También aplican las funciones len, min, max y sum (cuando los tipos lo permiten)
```

### 2.3 Métodos de una tupla

Al ser inmutable, una tupla solo ofrece dos métodos: `count` e `index`
(los mismos que en las listas, pero sin nada que modifique la tupla).

## 3. El módulo `datetime`

### 3.1 Creación de fechas

```python
import datetime

fecha1 = datetime.date(1982, 3, 26)
fecha2 = datetime.date(2026, 8, 18)

print(type(fecha1), fecha1)
print(type(fecha2), fecha2)
```

### 3.2 Diferencia entre fechas: `datetime.timedelta`

Al restar dos objetos `date` se obtiene un objeto `timedelta`, cuyos atributos
de interés son: `days`, `seconds`, `microseconds`, `total_seconds()`.

```python
dt1 = fecha2 - fecha1

print(type(dt1))
print("Diferencia          :", dt1)
print("Días transcurridos  :", dt1.days)
print("Años aproximados    :", round(dt1.days / 365, 2))
```

## 4. El módulo `numpy`

`numpy` agrega el tipo `ndarray` (arreglos n-dimensionales) y varios submódulos
de interés:

- **`random`** — generación de números aleatorios
- **`linalg`** — álgebra lineal (determinante, inversa, autovalores, `solve`, `svd`, ...)
- **`fft`** — transformada rápida de Fourier

```python
import numpy as np

print("Versión de numpy:", np.__version__)
```

### 4.1 Submódulo `linalg`: funciones más usadas

`det`, `inv`, `eig`, `eigh`, `matrix_rank`, `solve`, `svd`, `norm`, `qr`, `cholesky`.

> Para ver la documentación completa de cualquier función usar `help(np.linalg.<funcion>)`
> y para listar todo lo disponible en un módulo, `dir(np.linalg)`.

```python
# Ejemplo rápido de álgebra lineal con numpy
A = np.array([[4, 2],
              [1, 3]])

print("Matriz A:\n", A)
print("Determinante :", np.linalg.det(A))
print("Inversa      :\n", np.linalg.inv(A))
print("Autovalores  :", np.linalg.eigvals(A))
```
