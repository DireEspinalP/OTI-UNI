## Metodo `capitalize`

Es propio (y exclusivo) de una variable de tipo `str`. La documentacion de python esta escrita de manera precisa y sin ambiguedades.

```python
help("".capitalize)
```

```python
# Ejemplos
"pit705".capitalize()
```

```python
cad6 = "buenos dias, bienvenidos a la tercera sesion"
cad7 = cad6.capitalize()
cad7
```

## Metodo `isalnum`

```python
help(cad6.isalnum)
```

```python
# Ejemplos: cadena vacia, solo simbolos, alfanumerico, con espacios/simbolos
"".isalnum(), "#$#[]{}@".isalnum(), cad1.isalnum(), "666PIT".isalnum(), "11/08/2026".isalnum()
```
