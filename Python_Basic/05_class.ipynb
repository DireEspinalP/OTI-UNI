# Funcion de entrada : `input`

````python
# Ejemplo: area de un triangulo dado base y altura
base_triangulo = float(input("Ingresa la base del triangulo : "))
altura_triangulo = float(input("Ingresa la altura del triangulo : "))
area_triangulo = base_triangulo * altura_triangulo / 2
area_triangulo
````

````python
# Ejemplo: area de un triangulo dados sus 3 lados (Formula de Heron)
lado1 = float(input("Ingresa el primer lado del triangulo : "))
lado2 = float(input("Ingresa el segundo lado del triangulo : "))
lado3 = float(input("Ingresa el tercer lado del triangulo : "))

semiperimetro = (lado1 + lado2 + lado3) / 2
area = (semiperimetro * (semiperimetro - lado1) * (semiperimetro - lado2) * (semiperimetro - lado3)) ** 0.5
area
````

````python
# Ejemplo: aproximacion de seno(x) con el polinomio de Taylor
# seno(x) approx x - x^3/3! + x^5/5! - x^7/7!
import math as m

angulo = float(input("Ingresa el angulo (°): "))
x = angulo * m.pi / 180
seno_taylor = x - (x**3) / m.factorial(3) + (x**5) / m.factorial(5) - (x**7) / m.factorial(7)
seno_taylor
````

````python
# Comparemos el error de nuestra aproximacion vs. el modulo math (valor teorico: 0.5)
error1 = m.fabs(0.5 - seno_taylor)
error2 = m.fabs(0.5 - m.sin(m.radians(30)))
error1, error2

# El error del modulo math es menor: su aproximacion es mas precisa
````

# La funcion de salida : `print`

````python
# print() permite mostrar varios valores, con texto explicativo y saltos de linea (\n)
print("El error cometido por mi aprox. de taylor ", error1,
      "\nEl error cometido por la funcion seno del modulo math es  : ", error2)
````

````python
# Ejemplo: interes simple de un monto M0 a una tasa i del 10% anual, tiempo t
M0 = float(input("Ingresa el monto inicial : "))
t = float(input("Ingresa el tiempo (resolucion anual) : "))

i = 0.1
interes = M0 * i * t

print("El interes generado es: ", interes)
print("El interes generado es %.2f" % interes)
````

````python
# Otra forma de presentar resultados: docstring con formato
print("""
Calculadora de Interes
----------------------

    Monto Inicial : %.2f
    Tiempo : %.2f
    ====================

    Interes generado  : %.2f

""" % (M0, t, interes))
````

````python
# Ejemplo: segun la edad ingresada, se sugiere una accion
#   edad < 18            -> preparar el terreno para la siembra
#   18 <= edad < 29       -> sembrar
#   29 <= edad < 45       -> cuidar el terreno para futuras siembras
#   45 <= edad < 99       -> disfrutar la cosecha

edad_user = int(input("Ingresa tu edad (Numero Entero) : "))

if (edad_user >= 0) and (edad_user < 18):
    print("Es momento de preparar el terreno para la siembra")
elif (edad_user >= 18) and (edad_user < 29):
    print("Es momento de sembrar")
elif (edad_user >= 29) and (edad_user < 45):
    print("Es momento de cuidar el terreno para futuras siembras")
elif (edad_user >= 45) and (edad_user < 99):
    print("Es momento de disfrutar la cosecha")
else:
    print("El valor ingresado no es una edad adecuada (considerada)")

# Si el usuario ingresa algo que no sea un digito, int() lanza un ValueError
````

````python
# Version robusta: validamos con isdigit() antes de convertir a int
edad_user = input("Ingresa tu edad (Numero Entero) : ")

if edad_user.isdigit():
    edad_user = int(edad_user)
    if (edad_user >= 0) and (edad_user < 18):
        print("Es momento de preparar el terreno para la siembra")
    elif (edad_user >= 18) and (edad_user < 29):
        print("Es momento de sembrar")
    elif (edad_user >= 29) and (edad_user < 45):
        print("Es momento de cuidar el terreno para futuras siembras")
    elif (edad_user >= 45) and (edad_user < 99):
        print("Es momento de disfrutar la cosecha")
    else:
        print("El valor ingresado no es una edad adecuada (considerada)")
else:
    print("Ingreso ERRONEO")
````
