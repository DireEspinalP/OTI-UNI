# Definicion y operaciones con variables de naturaleza numerica

```python
# El nombre de una variable debe ser coherente con su uso
peso = 116      # kilo
estatura = 1.75 # metro
```

```python
# type() nos dice el tipo de dato de una variable
type(peso), type(estatura)
```

```python
# Python es de tipado dinamico: una variable puede cambiar de tipo
edad = 44
print(type(edad))
edad = "cuarenta y cuatro"
print(type(edad))
```

## Operaciones con variables de naturaleza numerica

- `+` suma  `-` resta  `*` multiplicacion  `/` division  `**` potenciacion
- `//` cociente entero  `%` residuo

Algoritmo de Euclides: `D = d * q + r`  (q = D // d, r = D % d)

```python
# Ejemplo: cuadrado de lado 8.5cm, comparado con un area de 60cm2
lado = 8.5
area = lado ** 2
area, area - 60
```

## El modulo math

```python
import math as m

# Tangente de 45°, de dos formas equivalentes
m.tan(45 * m.pi / 180), m.tan(m.radians(45))
```

```python
# Maximo comun divisor
m.gcd(13, 25)
```

## El modulo random

```python
import random as rnd

# Numero aleatorio con distribucion normal (media, desviacion estandar)
rnd.gauss(-2.3, 0.5)
```

# Variables de tipo Cadena de Caracter (str)

```python
s1 = "PIT705 Python Basico"
s2 = 'Segunda Sesion (06/08/2026)'
s3 = """Este es el peor curso de mi vida !!!"""
type(s1), type(s2), type(s3)
```

## Operaciones con variables de tipo str

```python
# Concatenacion (+) y repeticion (*)
s4 = s1 + " " + s2
s5 = 20 * s1
s6 = s2 * 26
type(s4), type(s5), type(s6)
```

```python
# Combinando concatenacion y repeticion
s7 = 19 * s6 + s5 * 82
type(s7), len(s7)
```

```python
# Indexado: primera, segunda, ultima, penultima, antepenultima componente
s1[0], s1[1], s1[-1], s1[-2], s1[-3]
```

```python
# Slicing: primeras 5 y ultimas 6 componentes
s3[:5], s3[-6:]
```

```python
# len() devuelve el numero de caracteres
len(s1)
```

```python
# Acceder fuera de rango produce IndexError
s1[29]
```
