# El objeto `self`

```python
# El modelo explicito : Python

class Persona:
    def __init__(self, nombre):
        self.nombre = nombre

    def saludar(self):
        # "self" es simplemente una variable local que apunta hacia la instancia
        print(f"Hola, soy {self.nombre}")
```

```python
# Uso 1
p = Persona("Eugene")
```

```python
# tipo de dato
type(p)
```

```python
# Descubrir la lista de metodos que se puede aplicar a un objeto instancia
# de la clase Persona
dir(p)
```

```python
# Esto es lo que nosotros escribimos
p.saludar()
```

```python
# Esto es lo que python ejecuta internamente
Persona.saludar(p)
# Notar que aca estamos llamando a un metodo desde la clase
```

```python
# Probar que self es __init__, self en un metodo y la variable externa que guarda la instancia
# apuntan al exacto mismo objetok en la memoria ram

class RastreadorDeMemoria:
    def __init__(self, valor):
        print(f"1. Dentro del __init__ : El parametro self tiene el ID {id(self)}")
        self.valor = valor

    def mostrar_id(self):
        print(f"2. Dentro del metodo : el parametro self tiene el ID {id(self)}")
```

```python
# Creamos una instancia de esta clase : RastreadorDeMemoria
print("Creando una instancia de la clase : RasteradorDeMemoria")
objeto = RastreadorDeMemoria(666)
```

```python
print(f"Fuera de la clase : La variable 'objeto' tiene el ID {id(objeto)} \n\n\n")
```

```python


print("Llamando al metodo ...")
objeto.mostrar_id()
```

# El Encapsulamiento

Niveles de Visibilidad

- Publico (`atributo`)
- Protegido (`_atributo`)
- Privado (`__atributo`)
- Propiedades (`@property`)

```python
# Ejemplito1 : Atributos Publicos vs. Atributos Privados
# Clase que represente un parametros fisico de un sensor.
# El valor actual puede leerse, pero el historial de cambios es un detalle de implementacion
# que no debe ser manipulado externamente.

class ParametroFisico:
    def __init__(self,valor_inicial: float):
        self.valor = valor_inicial      # publico
        self.__historial = []           # Privado (name mangling)

    def actualizar(self, nuevo_valor: float):
        self.__historial.append(self.valor)     # Logica interna controlada
        self.valor = nuevo_valor
```

```python
# Uso
p = ParametroFisico(0)
```

```python
p.actualizar(2.5)
```

```python
import random as rnd
for nm in range(5):
    p.actualizar(rnd.randint(0,9))
```

```python
print(p.__historial)
```

```python
# Ejemplito2 : Atributos protegidos y herencia
# Clase base para modelos estadisticos donde las subuclases necesitan acceso a
# configuraciones internas, pero el usuario final no deberia modificarlas directamente

class ModelBase:
    def __init__(self):
        self._hiperparametros = {}          # protegido : Contrato para subclases

class ModeloAvanzado(ModelBase):
    def configurar(self):
        # Acceso permitido y esperado desde la subclase
        self._hiperparametros["learning_rate"] = 1e-5
        return self._hiperparametros
```

```python
# Uso
m = ModeloAvanzado()
config = m.configurar()
```

```python
m._hiperparametros
```

```python
# Ejemplito4 : Validacion de datos con @property
# Aplicacion : Axiomas de Probabilidad
class DistribucionProbabilidad:
    def __init__(self, p:float):
        self._p = 0.0
        self.p = p      # Invoca al setter para validar desde inicio

    @property
    def p(self) -> float:
        return self._p

    @p.setter
    def p(self, valor: float):
        if not 0.0 <= valor <= 1:
            raise ValueError("La probabilidad debe ser exactamente en el intervalor [0,1]")
        self._p= valor
```

```python
d = DistribucionProbabilidad(0.5)
d.p
```

```python
d1 = DistribucionProbabilidad(1.1)
d1.p
```

```python
d1.p =-0.5
```

```python
# Ejemplito5 : Metodos protegidos para el patron template method
class pipelineETL:
    def ejecutar(self, datos: list) -> int:
        # Metodo Publico: Orquesta el proceso (template)
        datos = self._extraer(datos)
        datos = self._transformar(datos)
        return self._cargar(datos)

    def _extraer(self, datos:list) -> list:         # Protegido
        return datos.copy()

    def _transformar(self, datos: list)-> list:      # Protegido
        return[x**2 for x in datos]

    def _cargar(self, datos:list) -> int:            # Protegido
        return sum(datos)
```

```python
p=pipelineETL()
```

```python
resultado = p.ejecutar({1,2,3})
```

```python
# Ejemplito6
# Atributos de solo lectura (Inmutabilidad logica)

import hashlib
class TransaccionBlockChain:
    def __init__(self, monto:float, emisor:str, receptor:str):
        self._monto = monto
        self._hash = self._generar_hash(emisor, receptor, monto)

    def _generar_hash(self,emisor, receptor, monto) -> str:
        data = f"{emisor}{receptor}{monto}".encode()
        return hashlib.sha256(data).hexdigest()

    @property
    def hash(self)->str:
        return self._hash

    @property
    def monto(self)-> float:
        return self._monto
```

```python
t = TransaccionBlockChain(100, "Alice", "Bob")
```
