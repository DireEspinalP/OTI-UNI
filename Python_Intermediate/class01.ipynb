# Errores y excepciones en Python

* **Error**: falla que impide ejecutar el código u obtener resultados correctos.
* **Excepción**: evento en tiempo de ejecución que interrumpe el flujo, pero se puede capturar y manejar.

Tipos de errores: **sintaxis**, **lógicos** (semánticos) y **de ejecución** (excepciones).

```python
# Cada fragmento falla a propósito; se ejecuta aquí y se muestra el error en una línea
errores = {
    "sintaxis: falta ':'":      "if 3.14 > 1.2\n    print('pi')",
    "sintaxis: paréntesis":     "r = (23 + 12 * (9 * 3.21)",
    "sintaxis: palabra clave":  "def = 10",
    "indentación":              "if True:\nprint('Hola')",
    "ZeroDivisionError":        "10 / 0",
    "FileNotFoundError":        "open('datos.csv')",
    "KeyError":                 "{'nombre': 'Rosa'}['ciudad']",
    "IndexError":               "[12, 9, 3][6]",
    "ValueError":               "int('veinte')",
}
for nombre, src in errores.items():
    try:
        exec(compile(src, "<demo>", "exec"))
    except Exception as e:
        print(f"{nombre:26} -> {type(e).__name__}: {e}")
```

## Errores lógicos
No lanzan excepción, pero dan un resultado incorrecto.

```python
def promedio_mal(lista):            # resta en vez de sumar y resta 1 al final
    suma = 0.0
    for x in lista:
        suma -= x
    return suma / len(lista) - 1

def promedio(lista):
    return sum(lista) / len(lista)

def duplicar_mal(v): return v * 2            # repite la lista: [1,2,3,1,2,3]
def duplicar(v):     return [x * 2 for x in v]

edad = 16                                    # la condición debía ser >= 18
print("Puede votar" if edad >= 18 else "No puede votar")
print(promedio_mal([12, 14, 16]), promedio([12, 14, 16]))
print(duplicar_mal([1, 2, 3]), duplicar([1, 2, 3]))
```

## `try` · `except` · `else` · `finally`

* `try`: código que puede fallar
* `except`: maneja excepciones específicas
* `else`: solo si **no** hubo excepción
* `finally`: limpieza, se ejecuta siempre

```python
def dividir(a, b):
    try:
        resultado = a / b
    except ZeroDivisionError:
        print("Error: el denominador no puede ser cero")
        return None
    except TypeError:
        print("Error: a y b deben ser números")
        return None
    else:
        print(f"{a} / {b} = {resultado}")
        return resultado
    finally:
        print("Limpieza completada")

dividir(10, 2); dividir(10, 0); dividir("10", 2)
```

```python
def leer_numeros(ruta, columnas=5):
    """Lee un CSV numérico; devuelve [] si el archivo no existe o hay datos inválidos."""
    try:
        with open(ruta, encoding="utf-8-sig") as f:
            filas = [[float(c) for c in l.split(",")] for l in map(str.strip, f) if l]
        if any(len(fila) != columnas for fila in filas):
            raise ValueError(f"Todas las filas deben tener {columnas} columnas")
        return filas
    except FileNotFoundError:
        print("Error: no se encontró el archivo")
    except ValueError as e:
        print("Error: contenido inválido ->", e)
    return []

print(leer_numeros("no_existe.csv"))
```

## `assert`: documentación ejecutable
Valida precondiciones e invariantes.

```python
def raiz_cuadrada(x):
    assert x >= 0, "x debe ser no negativo"
    return x ** 0.5

def fibonacci(n):
    assert isinstance(n, int) and n >= 0, "n debe ser entero no negativo"
    seq = [1, 1][:n]
    for i in range(2, n):
        seq.append(seq[-1] + seq[-2])
    return seq

print(raiz_cuadrada(9), fibonacci(8))
for f, arg in [(raiz_cuadrada, -5.6), (fibonacci, -8), (fibonacci, 2.3)]:
    try:
        f(arg)
    except AssertionError as e:
        print(f"{f.__name__}({arg}) -> AssertionError: {e}")
```
